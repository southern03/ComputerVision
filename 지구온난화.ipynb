```python
import pandas as pd
import matplotlib.pyplot as plt

# 데이터 읽기
data = pd.read_excel('/content/시도별 평균기온.xlsx')

# 시도별로 데이터 그룹화
grouped_data = data.groupby('시도별')[['2000년', '2020년']].mean()

# 그래프 크기와 스타일 조정
plt.figure(figsize=(15, 5))

# 그래프 그리기
grouped_data.plot(kind='bar', figsize=(12, 6))
plt.title('시도별 평균 기온 (2000년 vs 2020년)')
plt.xlabel('시도별')
plt.ylabel('평균 기온')
plt.legend(title='연도', labels=['2000년', '2020년'])
plt.ylim(10, None)  # y 축 범위 설정 (10도부터 시작)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 데이터 읽기
data = pd.read_excel('/content/시도별 평균기온.xlsx')

# 시도별로 데이터 그룹화
grouped_data = data.groupby('시도별')[['2000년', '2020년']].mean()

# 그래프 크기와 스타일 조정
plt.figure(figsize=(15, 10))

# 세로 막대 그래프 그리기
grouped_data.plot(kind='barh', figsize=(12, 6))
plt.title('시도별 평균 기온 (2000년 vs 2020년)')
plt.xlabel('평균 기온')
plt.ylabel('시도별')
plt.legend(title='연도', labels=['2000년', '2020년'])
plt.xlim(10, None)  # x 축 범위 설정 (10도부터 시작)
plt.show()
```
